```python
f = open('060B-Working_Business_Proposal.pdf', 'rb')
```

```python
pdfr = pypdf.PdfReader(f)
```

```python
ls = []

for page in pdfr.pages:
    pg_text = page.extract_text()
    pg_txt = pg_text.split(' \n')
    
    for line in pg_txt:
        ls.append(line) 
    ls.append('\n')
ls
```

```python
f.close()
```
