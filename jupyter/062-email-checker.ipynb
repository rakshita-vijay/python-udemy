```python
smtp_ob = smtplib.SMTP('smtp.gmail.com', port=587)
# port can also be 465, or not passing the argument port at all

smtp_ob.ehlo() # greets the server and establishes the connection
```

```python
# port=587 : we are using TLS encryption
# port=465 : we are using SSL encryption
```
